```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
```

```python
req = requests.get('https://quotes.toscrape.com/')

with open('htmls\home.html', 'w', encoding="utf-8") as f:
    f.write(req.text)
```

```python
topicList = []
with open('htmls\home.html', 'r', encoding="utf-8") as f:
    content = f.read()
soup = BeautifulSoup(content , 'html.parser')

topics = soup.select_one("div.tags-box").find_all('span')
for span in topics:
   topicList.append(span.get_text(strip=True))

topicList
```

```python
for i in topicList:
    for k in range(1,3):
        req = requests.get(f"https://quotes.toscrape.com/tag/{i}/page/{k}/")
        with open(f'htmls\{i}page_{k}.html', 'w', encoding="utf-8") as f:
            f.write(req.text)
        
```

```python
row =[]
for i in topicList:
    for j in range(1,3):
        with open(f"htmls\{i}page_{j}.html", 'r', encoding="utf-8") as f:
            content = f.read()
        soup = BeautifulSoup(content , 'html.parser')
        quote = [q.find("span").text for q in soup.select("div.quote")]
        author = [q.find("small").text for q in soup.select("div.quote")]
        for q ,a in zip(quote, author):
            row.append([q,a])

df= pd.DataFrame(row , columns=['quote', 'author'])
```
