```python
from  bs4 import BeautifulSoup
import pandas as pd
```

```python
lst = []
for i in range(1,51):
    with open(f"htmls\page-{i}.html",  "r", encoding="utf-8") as f:
        content = f.read()
    soup = BeautifulSoup(content , 'html.parser')

    
    articles =soup.find_all("article")
    for article in articles:
        title =article.find('h3').find('a')['title']
        price = article.select_one('p.price_color').text.split('£')[1]
        rating = article.find('p')['class'][1]
        print(rating)
        lst.append([title, price, rating])




```

```python
df = pd.DataFrame(lst, columns=["Name", "price", "rating"])
```

```python
df.to_csv('final.csv', index=False)
```
