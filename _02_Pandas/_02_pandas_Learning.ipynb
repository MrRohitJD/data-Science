```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 14, 12, 18, 20]

# Plot
plt.figure(figsize=(6,4))                 # Width, Height
plt.plot(
    x, y, 
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=8,
    label="Sales Growth"
)

plt.title("Monthly Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
categories = ["A", "B", "C", "D"]
values = [20, 35, 30, 25]

# Bar Plot
plt.bar(
    categories, values,
    color="skyblue",
    width=0.5,
    edgecolor="black",
    linewidth=2,
    alpha=0.8,
    label="Students Count"
)

plt.title("Class-wise Students Data")
plt.xlabel("Class")
plt.ylabel("Number of Students")

plt.legend()
plt.grid(axis='y')    # horizontal grid lines

plt.show()
```
