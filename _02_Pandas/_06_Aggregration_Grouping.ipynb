# Aggregration & Grouping

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
```

```python
df
```

```python
df.groupby("Department")["Salary"].count()
```

```python
df.groupby('Team')['Gender'].count()
```

```python
df.groupby('Department')['Salary'].sum().shape
```

```python
df.groupby('Gender')['Gender'].count()
```

```python
#df.groupby('Department')['Salary'].aggregate( Addition = "sum", Mean_is = 'mean' , maximun = 'max')
df.groupby('Department')['Salary'].agg( Addition = "sum", Mean_is = 'mean' , maximun = 'max')
```

```python
df['Team Avg'] = df.groupby('Team')['Salary'].transform("mean")
df
```

```python
df.groupby().filter(lambda x: x["Salary"].mean() > 80)
```

```python
df.groupby("Team").filter(lambda x : x['Salary'].mean() >85)
```
