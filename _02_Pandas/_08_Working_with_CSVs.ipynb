## Working with CSVs

```python
import pandas as pd
```

```python
df = pd.read_csv('studentData1.csv',encoding='latin1')
```

```python
df1 = df[df['Marks']>90].copy()
```

```python
df1
```

```python
df1.to_csv('Marks90.csv', index=False)  # index=False remove the index columns
```

```python
df = pd.read_excel('studentData1.xlsx')
df
```

```python
df1 = df[df['Marks']<60]
```

```python
df1.to_excel('Marks35.xlsx', index=False)
```

```python
df1     
```

```python
df2 = df[df['Marks']<92]
```

```python
with pd.ExcelWriter('report.xlsx') as writer:
    df1.to_excel(writer, sheet_name='sheetZero1', index=False)
    df2.to_excel(writer, sheet_name='sheetZero2', index=False)
    
```
