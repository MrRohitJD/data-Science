## Merging & Joining Data

```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

```python
pd.merge(employees , departments,  on='DeptID')  # inner join
```

```python
pd.merge(employees , departments,  on='DeptID' , how= 'right')
```

```python
pd.merge(employees , departments , on='DeptID', how='left')
```

```python
pd.merge(employees, departments , on='DeptID', how='outer')
```

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"], "score" : [10 ,40]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"] , 'Age' : [26,52]})

```

```python
df1
```
